# Key Findings and Recommendations

## Key Findings

1. HHS discharge activity showed high variability, with a mean of 173.41
   children per day, a median of 181, and a coefficient of variation of
   72.49%.

2. HHS experienced 250 potential bottleneck days compared with only 20
   potential CBP bottleneck days, indicating greater operational pressure
   within the HHS stage of the pipeline.

3. CBP backlog pressure remained predominantly negative throughout the
   study period, while HHS backlog pressure showed greater fluctuations
   and became positive during several periods in 2024.

4. Cumulative CBP pressure reached approximately -32,212 by the end of
   the study period, indicating that transfers generally exceeded
   apprehensions over the period measured.

5. HHS discharge activity experienced a substantial decline during early
   2025, accompanied by a significant reduction in the monthly outcome
   stability score.

6. HHS care reached its highest observed level of 11,516 children on
   2023-12-20 and its lowest observed level of 1,972 children on
   2025-08-21.

7. Overall system activity declined considerably during 2025, with
   apprehensions, transfers and discharges reaching much lower levels
   than those observed during 2023–2024.

8. The 7-day moving-average analysis shows that sustained pressure can
   differ substantially from individual daily fluctuations, making
   rolling indicators useful for operational monitoring.

## Recommendations

1. Prioritize HHS capacity monitoring because of the substantially higher
   number of potential bottleneck days.

2. Use 7-day moving averages as an early-warning indicator for sustained
   operational pressure.

3. Monitor monthly outcome stability alongside volume-based KPIs.

4. Continuously compare CBP apprehensions with transfers to identify
   changes in custody pressure.

5. Investigate periods of major changes in discharge volume to understand
   whether they represent operational improvements, capacity changes,
   or changes in overall system activity.
